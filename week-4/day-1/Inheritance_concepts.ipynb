# Class and Objects

```python
class Student:
	def __init__(self,name,register_number):
		self.name = name
		self.register_number = register_number

	def display(self):
		print(f"Name : {self.name} and Register Number : {self.register_number}")
s1 = Student("Naveen","24CSEF013")
s1.display()
```

# __Inheritance__

### *Single Inheritance*

```python
class Animal:
	def __init__(self):
		print("Animal class called")

class Dog(Animal):
    def show(self):
        print("Dog class called")

dog = Dog()
dog.show()
```

### Multiple Inheritance

```python
class Father:
	def show(self):
		print("Father class called")

class Mother:
	def show(self):
		print("Mother class called")

class Son(Father,Mother):
	def show(self):
		super().show()
		print("Son class called")

son = Son()
son.show()
```

### Multilevel Inheritance

```python
class Grandparent:
	def show(self):
		print("Grandparent called")

class Parent(Grandparent):
	def show(self):
		super().show()
		print("Parent called")

class Daughter(Parent):
	def show(self):
		super().show()
		print("Daughter called")

daughter = Daughter()
daughter.show()
```

### Hierarchical Inheritance

```python
class Parent:
	def __init__(self):
		print("Parent called")

class Child1(Parent):
	def show(self):
		print("Child1 called")

class Child2(Parent):
	def show(self):
		print("Child2 called")

child1 = Child1()
child1.show()
child2 = Child2()
child2.show()
```

### Hybrid Inheritance

```python
class Vehicle:
	def __init__(self):
		print("Vehicle class Called")

class Car(Vehicle):
	def show():
		print("Car called")

class Bike(Vehicle):
	def show():
		print("Bike called")

class Walk(Car , Bike):
	def show():
		print("Walk called")

walk = Walk()
```

### Super Keyword 

```python
class A:
	def __init__(self):
		print("Class A  Init ")

class B(A):
	def __init__(self):
		print("Class B  Init ")
		super().__init__()

b = B()
```
